# Model Validation

## Binary Classification Evaluation

* Using model implementations of `sklearn` (or other) for Logistic Regression and SVM for classification tasks. Training both models on the `Breast Cancer` dataset.
* Evaluating the performance of both models using appropriate classification metrics and implementing them using `numpy` only.
 Reporting on the performance metrics: accuracy, precision, recall, F1-score.

````python
# Useful imports
import numpy as np
from sklearn.datasets import load_breast_cancer, load_diabetes, load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC, SVR
from sklearn.linear_model import LinearRegression, LogisticRegression
import matplotlib.pyplot as plt
````

````python
data = load_breast_cancer()
X = data.data
y = data.target

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)

scaler = StandardScaler()
X_tr_s = scaler.fit_transform(X_tr)
X_te_s = scaler.transform(X_te)

#LR
log_reg = LogisticRegression(max_iter=2000)
log_reg.fit(X_tr_s, y_tr)
y_pred_lr = log_reg.predict(X_te_s)

# SVM model
svm = SVC(kernel="rbf", probability=True)
svm.fit(X_tr_s, y_tr)
y_pred_svm = svm.predict(X_te_s)

def compute_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    TP = np.sum((y_pred == 1) & (y_true == 1))
    TN = np.sum((y_pred == 0) & (y_true == 0))
    FP = np.sum((y_pred == 1) & (y_true == 0))
    FN = np.sum((y_pred == 0) & (y_true == 1))

    accuracy = (TP + TN) / len(y_true)
    precision = TP / (TP + FP + 1e-10)
    recall = TP / (TP + FN + 1e-10)
    f1 = 2 * precision * recall / (precision + recall + 1e-10)

    return accuracy, precision, recall, f1

acc_lr, pre_lr, rec_lr, f1_lr = compute_metrics(y_te, y_pred_lr)
acc_svm, pre_svm, rec_svm, f1_svm = compute_metrics(y_te, y_pred_svm)

print("\n(RESULT) Logistic Regression Performance:")
print(f"Accuracy:  {acc_lr:.4f}")
print(f"Precision: {pre_lr:.4f}")
print(f"Recall:    {rec_lr:.4f}")
print(f"F1-score:  {f1_lr:.4f}")

print("\n(RESULT) SVM Performance:")
print(f"Accuracy:  {acc_svm:.4f}")
print(f"Precision: {pre_svm:.4f}")
print(f"Recall:    {rec_svm:.4f}")
print(f"F1-score:  {f1_svm:.4f}")
````

## Multi-Class Classification Evaluation

* Doing the same as for the multiclass problem `Iris`. Reporting on the performance metrics: accuracy, precision, recall, F1-score.


````python


from sklearn.datasets import load_iris

data = load_iris()
X = data.data
y = data.target

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)
X_tr_s = scaler.fit_transform(X_tr)
X_te_s = scaler.transform(X_te)

log_reg = LogisticRegression(max_iter=2000)
log_reg.fit(X_tr_s, y_tr)
y_pred_lr = log_reg.predict(X_te_s)

svm = SVC(kernel="rbf")
svm.fit(X_tr_s, y_tr)
y_pred_svm = svm.predict(X_te_s)

def compute_metrics_multiclass(y_true, y_pred):
    classes = np.unique(y_true)
    precision_list, recall_list, f1_list = [], [], []

    for c in classes:
        TP = np.sum((y_pred == c) & (y_true == c))
        FP = np.sum((y_pred == c) & (y_true != c))
        FN = np.sum((y_pred != c) & (y_true == c))
        precision = TP / (TP + FP + 1e-10)
        recall = TP / (TP + FN + 1e-10)
        f1 = 2 * precision * recall / (precision + recall + 1e-10)
        precision_list.append(precision)
        recall_list.append(recall)
        f1_list.append(f1)

    accuracy = np.sum(y_pred == y_true) / len(y_true)
    return accuracy, np.mean(precision_list), np.mean(recall_list), np.mean(f1_list)

acc_lr, pre_lr, rec_lr, f1_lr = compute_metrics_multiclass(y_te, y_pred_lr)
acc_svm, pre_svm, rec_svm, f1_svm = compute_metrics_multiclass(y_te, y_pred_svm)

print("\n(RESULT) Logistic Regression (Iris) Performance:")
print(f"Accuracy:  {acc_lr:.4f}")
print(f"Precision: {pre_lr:.4f}")
print(f"Recall:    {rec_lr:.4f}")
print(f"F1-score:  {f1_lr:.4f}")

print("\n(RESULT) SVM (Iris) Performance:")
print(f"Accuracy:  {acc_svm:.4f}")
print(f"Precision: {pre_svm:.4f}")
print(f"Recall:    {rec_svm:.4f}")
print(f"F1-score:  {f1_svm:.4f}")
````

## Task 5.1.3 - Regression Evaluation

* Now evaluating a trained `Linear Regression` and `SVM` model for the Regression task `Diabetes`. And, report on the performance metrics: MSE, RMSE, MAE, R².

````python
data = load_diabetes()
X = data.data
y = data.target

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, random_state=0)
scaler = StandardScaler()
X_tr_s = scaler.fit_transform(X_tr)
X_te_s = scaler.transform(X_te)

lin_reg = LinearRegression()
lin_reg.fit(X_tr_s, y_tr)
y_pred_lr = lin_reg.predict(X_te_s)

svm_reg = SVR(kernel="rbf")
svm_reg.fit(X_tr_s, y_tr)
y_pred_svm = svm_reg.predict(X_te_s)

def regression_metrics(y_true, y_pred):
    mse = np.mean((y_true - y_pred) ** 2)
    rmse = np.sqrt(mse)
    mae = np.mean(np.abs(y_true - y_pred))
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    r2 = 1 - ss_res / ss_tot
    return mse, rmse, mae, r2

mse_lr, rmse_lr, mae_lr, r2_lr = regression_metrics(y_te, y_pred_lr)
mse_svm, rmse_svm, mae_svm, r2_svm = regression_metrics(y_te, y_pred_svm)

print("\n(RESULT) Linear Regression (Diabetes) Performance:")
print(f"MSE:  {mse_lr:.4f}")
print(f"RMSE: {rmse_lr:.4f}")
print(f"MAE:  {mae_lr:.4f}")
print(f"R²:   {r2_lr:.4f}")

print("\n(RESULT) SVM Regression (Diabetes) Performance:")
print(f"MSE:  {mse_svm:.4f}")
print(f"RMSE: {rmse_svm:.4f}")
print(f"MAE:  {mae_svm:.4f}")
print(f"R²:   {r2_svm:.4f}")
````

## Cross-Validation

* Setting up a cross-validation pipeline for the `Linear Regression` and `SVM` models on the `Diabetes` dataset. (Regression)
* Setting up a cross-validation pipeline for the `Logistic Regression` and `SVM` models on the `Iris` dataset. (Classification)
* Reporting the performance metrics on all folds (minimum 5-fold) for each model and dataset.

````python
from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
import numpy as np

data = load_diabetes()
X = data.data
y = data.target

kf = KFold(n_splits=5, shuffle=True, random_state=0)

cv_results_reg = {"Linear Regression": [], "SVR": []}

for train_idx, test_idx in kf.split(X):
    X_tr, X_te = X[train_idx], X[test_idx]
    y_tr, y_te = y[train_idx], y[test_idx]


    scaler = StandardScaler()
    X_tr_s = scaler.fit_transform(X_tr)
    X_te_s = scaler.transform(X_te)

    lr = LinearRegression()
    lr.fit(X_tr_s, y_tr)
    pred_lr = lr.predict(X_te_s)

    svr = SVR(kernel="rbf")
    svr.fit(X_tr_s, y_tr)
    pred_svr = svr.predict(X_te_s)

    def reg_metrics(y_true, y_pred):
        mse = np.mean((y_true - y_pred) ** 2)
        rmse = np.sqrt(mse)
        mae = np.mean(np.abs(y_true - y_pred))
        ss_res = np.sum((y_true - y_pred)**2)
        ss_tot = np.sum((y_true - np.mean(y_true))**2)
        r2 = 1 - ss_res / ss_tot
        return mse, rmse, mae, r2

    cv_results_reg["Linear Regression"].append(reg_metrics(y_te, pred_lr))
    cv_results_reg["SVR"].append(reg_metrics(y_te, pred_svr))

print("\n(RESULT) Cross-Validation — Diabetes Regression")
for model, scores in cv_results_reg.items():
    print(f"\n{model}:")
    for i, (mse, rmse, mae, r2) in enumerate(scores):
        print(f" Fold {i+1}: MSE={mse:.4f}, RMSE={rmse:.4f}, MAE={mae:.4f}, R²={r2:.4f}")
````

````python
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

data = load_iris()
X = data.data
y = data.target

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)

cv_results_clf = {"Logistic Regression": [], "SVM": []}

for train_idx, test_idx in skf.split(X, y):
    X_tr, X_te = X[train_idx], X[test_idx]
    y_tr, y_te = y[train_idx], y[test_idx]

    scaler = StandardScaler()
    X_tr_s = scaler.fit_transform(X_tr)
    X_te_s = scaler.transform(X_te)

    lr = LogisticRegression(max_iter=2000)
    lr.fit(X_tr_s, y_tr)
    pred_lr = lr.predict(X_te_s)

    svm = SVC(kernel="rbf")
    svm.fit(X_tr_s, y_tr)
    pred_svm = svm.predict(X_te_s)

    def clf_metrics(y_true, y_pred):
        classes = np.unique(y_true)
        prec_list, rec_list, f1_list = [], [], []
        for c in classes:
            TP = np.sum((y_pred == c) & (y_true == c))
            FP = np.sum((y_pred == c) & (y_true != c))
            FN = np.sum((y_pred != c) & (y_true == c))
            precision = TP / (TP + FP + 1e-10)
            recall = TP / (TP + FN + 1e-10)
            f1 = 2 * precision * recall / (precision + recall + 1e-10)
            prec_list.append(precision)
            rec_list.append(recall)
            f1_list.append(f1)
        accuracy = np.sum(y_pred == y_true) / len(y_true)
        return accuracy, np.mean(prec_list), np.mean(rec_list), np.mean(f1_list)

    cv_results_clf["Logistic Regression"].append(clf_metrics(y_te, pred_lr))
    cv_results_clf["SVM"].append(clf_metrics(y_te, pred_svm))

print("\n\n(RESULT) Cross-Validation — Iris Classification")
for model, scores in cv_results_clf.items():
    print(f"\n{model}:")
    for i, (acc, prec, rec, f1) in enumerate(scores):
        print(f" Fold {i+1}: ACC={acc:.4f}, PREC={prec:.4f}, REC={rec:.4f}, F1={f1:.4f}")
````
