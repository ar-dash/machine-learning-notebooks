# Generative Models



## Flow Matching

* Implementing a small flow matching model and training loop.
* Training the flow matching model on the CIFAR10 dataset.


```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm

# Set random seed for reproducibility
torch.manual_seed(42)
np.random.seed(42)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
if device.type == 'cpu':
    print(" WARNING: Training on CPU will be slow. Consider using GPU or reducing epochs.")

# REDUCED Hyperparameters for faster training
batch_size = 256  # Larger batches = fewer iterations
learning_rate = 3e-4
num_epochs = 50
image_size = 32
channels = 3

# Data preprocessing for CIFAR-10
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

# Load CIFAR-10 dataset
train_dataset = datasets.CIFAR10(root='./data', train=True, transform=transform, download=True)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)

print(f"Dataset loaded: {len(train_dataset)} training images")

# LIGHTER U-Net architecture (fewer parameters)
class LightUNet(nn.Module):
    def __init__(self, in_channels=3, time_emb_dim=64):
        super(LightUNet, self).__init__()

        # Simpler time embedding
        self.time_mlp = nn.Sequential(
            nn.Linear(1, time_emb_dim),
            nn.SiLU()
        )

        # Lighter encoder (fewer channels)
        self.enc1 = self.conv_block(in_channels, 32)
        self.enc2 = self.conv_block(32, 64)

        # Bottleneck
        self.bottleneck = self.conv_block(64 + time_emb_dim, 128)

        # Lighter decoder
        self.dec2 = self.conv_block(128 + 64, 64)
        self.dec1 = self.conv_block(64 + 32, 32)

        # Final output
        self.final = nn.Conv2d(32, in_channels, kernel_size=1)

        self.pool = nn.MaxPool2d(2, 2)
        self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)

    def conv_block(self, in_ch, out_ch):
        # Single conv instead of double
        return nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1),
            nn.GroupNorm(8, out_ch),
            nn.SiLU()
        )

    def forward(self, x, t):
        # Time embedding
        t_emb = self.time_mlp(t.view(-1, 1))

        # Encoder (only 2 levels instead of 3)
        enc1 = self.enc1(x)                    # [B, 32, 32, 32]
        enc2 = self.enc2(self.pool(enc1))      # [B, 64, 16, 16]

        # Expand time embedding
        B, C, H, W = enc2.shape
        t_emb = t_emb.view(B, -1, 1, 1).expand(B, -1, H, W)

        # Bottleneck with time embedding
        bottleneck = self.bottleneck(torch.cat([enc2, t_emb], dim=1))  # [B, 128, 16, 16]

        # Decoder
        dec2 = self.dec2(torch.cat([bottleneck, enc2], dim=1))          # [B, 64, 16, 16]
        dec1 = self.dec1(torch.cat([self.upsample(dec2), enc1], dim=1)) # [B, 32, 32, 32]

        return self.final(dec1)

# Flow matching functions
def sample_time(batch_size):
    return torch.rand(batch_size, device=device)

def compute_flow_matching_loss(model, x1):
    batch_size = x1.shape[0]
    t = sample_time(batch_size)
    x0 = torch.randn_like(x1, device=device)
    x_t = t.view(-1, 1, 1, 1) * x1 + (1 - t.view(-1, 1, 1, 1)) * x0
    u_t = x1 - x0
    v_t = model(x_t, t)
    loss = torch.mean((v_t - u_t) ** 2)
    return loss

# Initialize lighter model
model = LightUNet(in_channels=channels).to(device)
optimizer = optim.AdamW(model.parameters(), lr=learning_rate)

print(f"\nModel parameters: {sum(p.numel() for p in model.parameters()):,}")
print(f"Training for {num_epochs} epochs with batch size {batch_size}")

# Training loop
losses = []

for epoch in range(num_epochs):
    model.train()
    epoch_loss = 0.0

    progress_bar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}')
    for data, _ in progress_bar:
        data = data.to(device)

        loss = compute_flow_matching_loss(model, data)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()
        progress_bar.set_postfix({'loss': f'{loss.item():.4f}'})

    avg_loss = epoch_loss / len(train_loader)
    losses.append(avg_loss)
    print(f'Epoch [{epoch+1}/{num_epochs}], Avg Loss: {avg_loss:.4f}')

print("\n Training completed!")

# Plot loss
plt.figure(figsize=(10, 5))
plt.plot(losses, marker='o')
plt.title('Flow Matching Training Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)
plt.savefig('flow_matching_loss.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Image Generation

* Visualizing the training process by showing intermediate samples during training.
* Sample 10 synthetic images using your final flow matching model.


```python
# Sampling function using ODE solver
@torch.no_grad()
def sample_images(model, num_samples=10, num_steps=100):
    """
    Sample images using the learned flow
    """
    model.eval()

    # Start from noise
    x = torch.randn(num_samples, channels, image_size, image_size, device=device)

    # Time steps for ODE integration
    dt = 1.0 / num_steps

    intermediate_samples = []
    save_steps = [0, num_steps//4, num_steps//2, 3*num_steps//4, num_steps-1]

    for i in range(num_steps):
        t = torch.ones(num_samples, device=device) * (i / num_steps)

        # Predict velocity
        v_t = model(x, t)

        # Euler step
        x = x + v_t * dt

        # Save intermediate samples
        if i in save_steps:
            intermediate_samples.append(x.cpu().clone())

    return x, intermediate_samples

# Generate samples
print("\nGenerating samples...")
final_samples, intermediate_samples = sample_images(model, num_samples=10, num_steps=100)

# Denormalize images
def denormalize(img):
    return (img * 0.5 + 0.5).clamp(0, 1)

# Visualize intermediate samples during generation
print("\nVisualizing generation process...")
n_steps = len(intermediate_samples)
n_samples = min(5, final_samples.shape[0])

fig, axes = plt.subplots(n_steps, n_samples, figsize=(15, 3*n_steps))

for step_idx, samples in enumerate(intermediate_samples):
    for img_idx in range(n_samples):
        img = denormalize(samples[img_idx]).permute(1, 2, 0).numpy()

        if n_steps == 1:
            ax = axes[img_idx]
        else:
            ax = axes[step_idx, img_idx]

        ax.imshow(img)
        ax.axis('off')

        if img_idx == 0:
            step_names = ['Start (Noise)', 'Step 25', 'Step 50', 'Step 75', 'Final']
            ax.set_title(step_names[step_idx], fontsize=10, pad=5)

plt.tight_layout()
plt.savefig('flow_matching_generation_process.png', dpi=150, bbox_inches='tight')
plt.show()

# Visualize final generated samples
print("\nVisualizing final synthetic images...")
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for idx in range(10):
    row = idx // 5
    col = idx % 5

    img = denormalize(final_samples[idx].cpu()).permute(1, 2, 0).numpy()
    axes[row, col].imshow(img)
    axes[row, col].axis('off')
    axes[row, col].set_title(f'Sample {idx+1}', fontsize=10)

plt.suptitle('Synthetic CIFAR-10 Images Generated by Flow Matching', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('flow_matching_samples.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nFlow matching task completed!")

# Compare with real images
print("\nComparing with real CIFAR-10 images...")
real_images, _ = next(iter(train_loader))
real_images = real_images[:10]

fig, axes = plt.subplots(2, 10, figsize=(20, 4))

# Real images
for idx in range(10):
    img = denormalize(real_images[idx]).permute(1, 2, 0).numpy()
    axes[0, idx].imshow(img)
    axes[0, idx].axis('off')
    if idx == 0:
        axes[0, idx].set_title('Real', fontsize=12, pad=10)

# Generated images
for idx in range(10):
    img = denormalize(final_samples[idx].cpu()).permute(1, 2, 0).numpy()
    axes[1, idx].imshow(img)
    axes[1, idx].axis('off')
    if idx == 0:
        axes[1, idx].set_title('Generated', fontsize=12, pad=10)

plt.tight_layout()
plt.savefig('flow_matching_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nAll visualizations completed!")
```
